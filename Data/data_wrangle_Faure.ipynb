```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# GRB2 SH3

```python
binding_data = pd.read_csv('GRB2-SH3_bind_all_mutants.csv', index_col=0)
binding_data = binding_data[['aa_seq', 'fitness']]
binding_data = binding_data.rename(columns = {'fitness' : 'activity'})

# Filtering out the sequence that appears many times (WT sequence maybe?)
# binding_data = binding_data[binding_data['aa_seq']  != 'TYVQALFDFDPQEDGELGFRRGDFIHVMDNSDPNWWKGACHGQTGMFPRNYVTPVN*']
binding_data = pd.DataFrame(binding_data.groupby('aa_seq')['activity'].mean()).reset_index()
binding_data
```

```python
abund_data = pd.read_csv('GRB2-SH3_abund_all_mutants.csv', index_col=0)
abund_data = abund_data[['aa_seq', 'fitness']]
abund_data = abund_data.rename(columns = {'fitness' : 'abundance'})

# Filtering out the sequence that appears many times (WT maybe?)
# abund_data = abund_data[abund_data['aa_seq'] != 'TYVQALFDFDPQEDGELGFRRGDFIHVMDNSDPNWWKGACHGQTGMFPRNYVTPVN*']
abund_data = pd.DataFrame(abund_data.groupby('aa_seq')['abundance'].mean()).reset_index()
abund_data
```

```python
len(set(binding_data['aa_seq'].str.removesuffix('*')).difference(set(abund_data['aa_seq'].str.removesuffix('*'))))
```

```python
merged_data = abund_data.merge(binding_data, on='aa_seq', how='inner')
merged_data['aa_seq'] = merged_data['aa_seq'].str.removesuffix('*')
merged_data = merged_data[['*' not in seq for seq in merged_data['aa_seq']]]
merged_data
```

```python
merged_data.to_csv("GRB2_SH3_wrangled.csv")
```
